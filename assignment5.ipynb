```python
#Q1)
numbers = [59, 15, 29, 55, 39, 78]
even = 0
odd = 0
for i in numbers:
    if i % 2 == 0:
        even += 1
    else:
      odd += 1
print("Even numbers:", even)
print("Odd numbers:", odd) 
```

```python
#Q2)
numbers = [-10, 20, -35, 40, 30, 15]
positive = []
for num in numbers:
    if num > 0:
      positive.append(num)
print(positive)
```

```python
#Q3)
numbers = [10, 20, 30, 20, 40, 10, 50]
result = []
for num in numbers:
    if num not in result:
        result.append(num)
print(result)
```

```python
#Q4)
numbers = [412, 415, 967, 889, 754]
numbers.sort()
print("Second Largest:", numbers[-2])
```

```python
#Q5)
numbers = [12, 45, 67, 89, 54]
numbers.sort()
print("Second Smallest:", numbers[2])
```

```python
#Q6)
numbers = [1, 2, 3, 2, 1, 4, 1]
frequency = {}
for num in numbers:
    if num in frequency:
        frequency[num] += 1
    else:
        frequency[num] = 1
for key, value in frequency.items():
    print(key, "->", value)
```

```python
#Q7)
numbers = [100, 270, 380, 640, 950]
reversed_list = []
for i in range(len(numbers) - 1, -1, -1):
    reversed_list.append(numbers[i])
print(reversed_list)
```

```python
#Q8)
numbers = [1, 2, 3, 2, 1]
if numbers == numbers[::-1]:
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
#Q9)
list1 = [170, 920, 307, 408]
list2 = [20, 40, 60]
common = []
for num in list1:
    if num in list2:
        common.append(num)
print(common)
```
