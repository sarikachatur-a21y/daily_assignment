1. Student Class – __str__() 
Create a Student class with name, roll_no, and marks. Implement __str__() so that

```python
class Student:
    def __init__(self, name, roll_no, marks):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks
    def __str__(self):
        return self.name + "   "+str(self.roll_no) + " " + str(self.marks)
s1 = Student("Sarika", 101, 85)
s2 = Student("Priya", 102, 90)
s3 = Student("Sneha", 103, 78)
print(s1)
print(s2)
print(s3)
```

2. Product Class – __eq__() 
Create a Product class with name and price. Implement __eq__() to check whether two 
products have the same price. 

```python
class Product:
    def __init__(self, name, price):
        self.name = name
        self.price = price
    def __eq__(self, other):
        return self.price == other.price
p1 = Product("Laptop", 50000)
p2 = Product("Mobile", 50000)
print(p1 == p2)
```

3. Shopping Cart – __len__() 
Create a ShoppingCart class containing a list of products. Implement __len__() so that 
len(cart) returns the number of products in the cart.

```python
class ShoppingCart:
    def __init__(self):
        self.products = []
    def add_product(self, product):
        self.products.append(product)
    def __len__(self):
        return len(self.products)
cart = ShoppingCart()
cart.add_product("Laptop")
cart.add_product("Mobile")
cart.add_product("Mouse")
print(len(cart))
```

Create a BankAccount class with account holder name, account number, and balance.  
Implement __str__() to display account information when the object is printed. 

```python
class BankAccount:
    def __init__(self, name, account_no, balance):
        self.name = name
        self.account_no = account_no
        self.balance = balance
        
    def __str__(self):
        return self.name + " " + str(self.account_no) + " " + str(self.balance)
account = BankAccount("Sarika", 12345, 50000)
print(account)
```

5.Number Class – __add__() 
Create a Number class that stores one number. Implement __add__() so that two 
objects can be added

```python
class Number:
    def __init__(self, num):
        self.num = num
    def __add__(self, other):
        return self.num + other.num
n1 = Number(106)
n2 = Number(270)
print(n1 + n2)
```

6. Create a Number class and implement: 
__add__()  
__sub__()  
__mul__()  
__truediv__()  
__mod__()

```python
class Number:
    def __init__(self, num):
        self.num = num
    def __add__(self, other):
        return self.num + other.num
    def __sub__(self, other):
        return self.num - other.num
    def __mul__(self, other):
        return self.num * other.num
    def __truediv__(self, other):
        return self.num / other.num
    def __mod__(self, other):
        return self.num % other.num
n1 = Number(25)
n2 = Number(4)
print("Addition:", n1 + n2)
print("Subtraction:", n1 - n2)
print("Multiplication:", n1 * n2)
print("Division:", n1 / n2)
print("Modulus:", n1 % n2)
```
