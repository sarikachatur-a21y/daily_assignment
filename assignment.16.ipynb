3.Create an Employee parent class with name and salary. Create a Manager child class  with an additional department attribute. Display all details using constructors.  

```python
class Employee:
    def __init__(self, name, salary):
        self.name = name
        self.salary = salary
class Manager(Employee):
    def __init__(self, name, salary, department):
        super().__init__(name, salary)
        self.department = department
m1 = Manager("Sarika", 50000, "IT")
print("Name-", m1.name)
print("Salary-", m1.salary)
print("Department-", m1.department)
```

9 Student Marks 
Create a Student class with private attributes __name and __marks.  
Create getter and setter methods for marks.  
Allow marks only between 0 and 100.  
Create a child class ExamStudent that inherits from Student 


```python
class Student:
    def __init__(self, name, marks):
        self.__name = name
        self.__marks = 0
        self.set_marks(marks)
    def get_marks(self):
        return self.__marks
    def set_marks(self, marks):
        if 0 <= marks <= 100:
            self.__marks = marks
        else:
            print("Invalid marks! Marks must be between 0 and 100.")
    def display(self):
        print("Name:", self.__name)
        print("Marks:", self.__marks)
class ExamStudent(Student):
    def display_exam(self):
        print("Exam Student Details:")
        self.display()
student = ExamStudent("Sarika", 85)
student.display_exam()
student.set_marks(95)
print("Updated Marks:", student.get_marks())
student.set_marks(120)
```

Create get_password() and set_password().  
Setter should update the password only if it contains at least 6 characters. 

```python
class User:
    def __init__(self, password):
        self.__password = password

    def get_password(self):
        return self.__password

    def set_password(self, password):
        if len(password) >= 6:
            self.__password = password
            print("Password updated successfully.")
        else:
            print("Password must contain at least 6 characters.")
user = User("abc123")
print("Password:", user.get_password())
user.set_password("python123")
print("Updated Password:", user.get_password())
user.set_password("abc")
```

10. Multilevel Inheritance + Getter/Setter 
Person → private age with getter/setter.  
Employee → salary.  
Manager → department.  
Use super() at each level

```python
class Person:
    def __init__(self, age):
        self.__age = age
    def get_age(self):
        return self.__age
    def set_age(self, age):
        if age > 0:
            self.__age = age
        else:
            print("Invalid age.")
class Employee(Person):
    def __init__(self, age, salary):
        super().__init__(age)
        self.salary = salary
class Manager(Employee):
    def __init__(self, age, salary, department):
        super().__init__(age, salary)
        self.department = department
    def display(self):
        print("Age-", self.get_age())
        print("Salary-", self.salary)
        print("Department-", self.department)
manager = Manager(30, 50000, "IT")
manager.display()
manager.set_age(35)
print("Updated Age:", manager.get_age())
```
