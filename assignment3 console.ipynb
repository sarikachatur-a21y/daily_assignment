```python
#Q1) Write a program to take a student's name, age, city, and percentage as input and display them.
name = input("Enter Name: ")
age = int(input("Enter Age: "))
city = input("Enter City: ")
percentage = float(input("Enter Percentage: "))

print("\nStudent Details")
print("Name:", name)
print("Age:", age)
print("City:", city)
print("Percentage:", percentage)
```

```python
#Q2)Take the length and width of a rectangle as input and calculate: 
# -Area
# - Perimeter 
length = float(input("Enter Length: "))
width = float(input("Enter Width: "))

area = length * width
perimeter = 2 * (length + width)

print("Area =", area)
print("Perimeter =", perimeter)
```

```python
#Q3)Take the price and quantity of a product from the user and calculate:
# - Total Amount 
# - GST (18%)
# _Final Bill 
price = float(input("Enter Price: "))
quantity = int(input("Enter Quantity: "))
total = price * quantity
gst = total * 0.18
final = total + gst
print("Total Amount =", total)
print("GST =", gst)
print("Final Bill =", final)
```

```python
#Q4)Take a student's marks  Python, Java, and SQL as input and calculate:
#total
# Average 
# Percentage 
python = int(input("Python Marks: "))
java = int(input("Java Marks: "))
sql = int(input("SQL Marks: "))
total = python + java + sql
average = total / 3
percentage = total / 3
print("Total =", total)
print("Average =", average)
print("Percentage =", percentage)
```

```python
#Q8)Create a simple calculator using match-case.
# - Operations: 
# • Addition 
# • Subtraction  
# • Multiplication
# • Division 
a = int(input("Enter First Number: "))
b = int(input("Enter Second Number: "))
op = input("Enter Operation (+,-,*,/): ")
match op:
  case "+":
        print("Result =", a + b)
  case "-":
        print("Result =", a - b)
  case "*":
        print("Result =", a * b)
  case "/":
        print("Result =", a / b)
  case _:
        print("Invalid Operation")
```

```python
#Q9) Write a program using the Ternary Operator to determine whether a number is Even or Odd. 
num = int(input("Enter Number: "))
print("Even" if num % 2 == 0 else "Odd")
```
