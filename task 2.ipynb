```python
numbers = [10, 20, 30, 40, 40, 50]
unique = []
for num in numbers:
    if num not in unique:
        unique.append(num)
largest = max(unique)
second_largest = None
for num in unique:
    if num != largest:
        if second_largest is None or num > second_largest:
            second_largest = num
print("Second largest:", second_largest)
```

```python
products1 = {
    "Laptop": 50000,
    "Mouse": 500
}

products2 = {
    "Laptop": 55000,
    "Keyboard": 1000
}

result = products1.copy()
for product, price in products2.items():
    if product in result:
        if price > result[product]:
            result[product] = price
    else:
        result[product] = price
print(result)
```

```python
def calculator(a, b, operation):
    if operation == "+":
        return a + b

    elif operation == "-":
        return a - b

    elif operation == "*":
        return a * b

    elif operation == "/":
        return a / b

    else:
        return "Invalid operation"
print(calculator(10, 5, "+"))
print(calculator(10, 5, "-"))
print(calculator(10, 5, "*"))
print(calculator(10, 5, "/"))
```

```python
def calculate(numbers):
    largest = max(numbers)
    smallest = min(numbers)
    total = sum(numbers)
    average = total / len(numbers)
    return largest, smallest, total, average
numbers = [10, 20, 30, 40, 50]
largest, smallest, total, average = calculate(numbers)
print("Largest:", largest)
print("Smallest:", smallest)
print("Sum:", total)
print("Average:", average)
```

```python
class BankAccount:

    def __init__(self, holder, account_number, balance):
        self.holder = holder
        self.account_number = account_number
        self.balance = balance

    def deposit(self, amount):
        self.balance += amount
        print("Deposited:", amount)

    def withdrawal(self, amount):

        if amount <= self.balance:
            self.balance -= amount
            print("Withdrawn:", amount)
        else:
            print("Insufficient balance")

    def display_balance(self):
        print("Account Holder:", self.holder)
        print("Account Number:", self.account_number)
        print("Balance:", self.balance)

account = BankAccount("Sarika", 12345, 10000)
account.display_balance()
account.deposit(5000)
account.withdrawal(3000)
account.display_balance()
```

```python
class Employee:
    def __init__(self, name, emp_id, department, salary):
        self.name = name
        self.emp_id = emp_id
        self.department = department
        self.salary = salary

    def display_details(self):
        print("Name:", self.name)
        print("ID:", self.emp_id)
        print("Department:", self.department)
        print("Salary:", self.salary)

    def increase_salary(self, percentage):
        self.salary = self.salary + (self.salary * percentage / 100)

    def check_salary(self):

        if self.salary > 50000:
            print("Salary is above ₹50,000")
        else:
            print("Salary is not above ₹50,000")


employee = Employee("Sarika", 101, "IT", 45000)

employee.display_details()

employee.increase_salary(20)

print("After salary increase:")
employee.display_details()

employee.check_salary()
```

```python
class Vehicle:

    def __init__(self, brand, speed):
        self.brand = brand
        self.speed = speed

    def display(self):
        print("Brand:", self.brand)
        print("Speed:", self.speed)


class Car(Vehicle):

    def __init__(self, brand, speed, doors):
        super().__init__(brand, speed)
        self.doors = doors

    def display(self):
        super().display()
        print("Doors:", self.doors)


class Bike(Vehicle):

    def __init__(self, brand, speed, engine):
        super().__init__(brand, speed)
        self.engine = engine

    def display(self):
        super().display()
        print("Engine:", self.engine)


car = Car("Toyota", 180, 4)
bike = Bike("Honda", 120, "150cc")

print("Car Details:")
car.display()

print("\nBike Details:")
bike.display()
```

```python
# Dictionary तयार करणे (Heterogeneous & Key-Value Pairs)
student = {
    "roll_no": 101,
    "name": "Rahul",
    "marks": 85.5,
    "passed": True
}

# Key चा वापर करून Value एक्सेस करणे (Indexed by Keys)
print(student["name"])  # Output: Rahul

# Mutable - व्हॅल्यू बदलणे किंवा नवीन जोडणे
student["marks"] = 90.0  # Update existing key
student["city"] = "Pune"  # Add new key-value pair

print(student)
```
