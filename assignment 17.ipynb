```python
#1. College Management — Aggregation Create Student and Address classes. 
# Address should contain city, state, and pincode.  
# Student should contain name and roll number.  
# A Student object should contain an Address object.  
# Display complete student information.  
# Demonstrate the HAS-A relationship.


 
class Address:
    def __init__(self, city, state, pincode):
        self.city = city
        self.state = state
        self.pincode = pincode
class Student:
    def __init__(self, name, roll_no, address):
        self.name = name
        self.roll_no = roll_no
        self.address = address
    def display(self):
        print("Name-", self.name)
        print("Roll No-", self.roll_no)
        print("City-", self.address.city)
        print("State-", self.address.state)
        print("Pincode-", self.address.pincode)
address = Address("Pune", "Maharashtra", 411001)
student = Student("Sarika", 101, address)
student.display()
```

2. Car & Engine — Aggregation 
Create Engine and Car classes. 
• Engine should have engine_no and cc.  
• Car should have model and price.  
• A Car should contain an Engine object.  
• Create a method to display car and engine details.  
• Start the engine through the Car object

```python
class Engine:
    def __init__(self, engine_no, cc):
        self.engine_no = engine_no
        self.cc = cc
    def start(self):
        print("Engine started")
class Car:
    def __init__(self, model, price, engine):
        self.model = model
        self.price = price
        self.engine = engine
    def display(self):
        print("Model-", self.model)
        print("Price-", self.price)
        print("Engine No-", self.engine.engine_no)
        print("CC-", self.engine.cc)
    def start_engine(self):
        self.engine.start()
engine = Engine("ENG169", 1280)
car = Car("enova", 8000000, engine)
car.display()
car.start_engine()
```
