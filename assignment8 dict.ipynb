```python
#Q1) Program to count vowels and consonants
string = input("Enter a string: ")
vowels = 0
consonants = 0
for char in string.lower():
    if char.isalpha():  # Check if the character is a letter
      if char in "aeiou": vowels += 1
    else:
        consonants += 1
print("Number of vowels:", vowels)
print("Number of consonants:", consonants)
```

```python
#Q2) Program to check whether a string is a palindrome
string = input("Enter a string: ")
reverse = string[::-1]# Reverse the string
if string == reverse:# Check if original and reversed strings are the sam
 print("Palindrome")
else:
 print("Not a Palindrome")
```

```python
#Q6)
data = {'a': 10, 'b': 20, 'c': 10, 'd': 30}
result = {}
seen = set()
for key, value in data.items():
   if value not in seen:
      result[key] = value
      seen.add(value)

print(result)
```
