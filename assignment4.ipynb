```python
#Q6)
sentence = input("Enter a sentence: ")
words = sentence.split()
longest = ""
for word in words:
    if len(word) > len(longest):
        longest = word
print("Longest Word:", longest)
```

```python
#Q7)
sentence = input("Enter a sentence: ")
result = ""
new_word = True
for ch in sentence:
    if new_word and ch.isalpha():
        result += ch.upper()
        new_word = False
    else:
        result += ch
    if ch == " ":
        new_word = True
print(result)
```

```python
#Q8)
sentence = input("Enter a sentence: ")
count = 0
in_word = False
for ch in sentence:
    if ch != " " and not in_word:
        count += 1
        in_word = True
    elif ch == " ":
        in_word = False
print("Total Words:", count)
```
