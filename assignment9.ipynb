```python
def find_largest_smallest(numbers):
    largest = max(numbers)
    smallest = min(numbers)
    return largest, smallest

numbers = [10, 25, 5, 40, 15]

largest, smallest = find_largest_smallest(numbers)

print("Largest =", largest)
print("Smallest =", smallest)
```

```python
def calculate(*args):
    total = sum(args)
    average = total / len(args)
    maximum = max(args)
    minimum = min(args)

    return total, average, maximum, minimum

result = calculate(10, 20, 30, 40, 50)

print("Sum =", result[0])
print("Average =", result[1])
print("Maximum =", result[2])
print("Minimum =", result[3])
```

```python
def count_characters(text):
    uppercase = 0
    lowercase = 0
    digits = 0
    for ch in text:
        if ch.isupper():
            uppercase += 1
        elif ch.islower():
            lowercase += 1
        elif ch.isdigit():
            digits += 1
    return uppercase, lowercase, digits
text = input("Enter a string: ")
upper, lower, digit = count_characters(text)
print("Uppercase letters =", upper)
print("Lowercase letters =", lower)
print("Digits =", digit)
```

```python
def check_palindrome(text):
    if text == text[::-1]:
        return True
    else:
        return False

text = input("Enter a string: ")

if check_palindrome(text):
    print("Palindrome")
else:
    print("Not Palindrome")
```

```python
def get_even_numbers(numbers):
    even_numbers = []

    for num in numbers:
        if num % 2 == 0:
            even_numbers.append(num)

    return even_numbers

numbers = [1, 2, 3, 4, 5, 6, 8, 9]

result = get_even_numbers(numbers)

print("Even numbers =", result)
```

```python
def addition(a, b):
    return a + b
def subtraction(a, b):
    return a - b

def multiplication(a, b):
    return a * b

def division(a, b):
    if b == 0:
        return "Cannot divide by zero"
    return a / b
while True:
    print("\n--- Calculator ---")
    print("1. Addition")
    print("2. Subtraction")
    print("3. Multiplication")
    print("4. Division")
    print("5. Exit")
    try:
        choice = int(input("Enter your choice: "))
        if choice == 5:
            print("Calculator closed.")
            break
        if choice in [1, 2, 3, 4]:
            a = float(input("Enter first number: "))
            b = float(input("Enter second number: "))
            if choice == 1:
                print("Result =", addition(a, b))
            elif choice == 2:
                print("Result =", subtraction(a, b))
            elif choice == 3:
                print("Result =", multiplication(a, b))
            elif choice == 4:
                print("Result =", division(a, b))
        else:
            print("Invalid choice! Please enter 1 to 5.")
    except ValueError:
        print("Invalid input! Please enter a number.")
```

```python
def find_topper(students):
    topper = max(students, key=students.get)
    highest_marks = students[topper]
    return topper, highest_marks
students = {
    "Rahul": 85,
    "Sneha": 92,
    "Amit": 78,
    "Priya": 95
}
topper, marks = find_topper(students)
print("Topper =", topper)
print("Highest Marks =", marks)
```

```python
def word_frequency(sentence):
    words = sentence.lower().split()
    frequency = {}

    for word in words:
        if word in frequency:
            frequency[word] += 1
        else:
            frequency[word] = 1
    return frequency
sentence = input("Enter a sentence: ")
result = word_frequency(sentence)
print("Word Frequency:")
for word, count in result.items():
    print(word, ":", count)
```

```python
def analyze_numbers(numbers):
    total = sum(numbers)
    even_count = 0
    odd_count = 0

    for num in numbers:
        if num % 2 == 0:
            even_count += 1
        else:
            odd_count += 1

    return total, even_count, odd_count


numbers = [10, 15, 20, 85, 37, 35]

total, even, odd = analyze_numbers(numbers)

print("Sum of numbers =", total)
print("Count of even numbers =", even)
print("Count of odd numbers =", odd)
```
