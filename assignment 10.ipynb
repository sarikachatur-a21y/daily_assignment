```python
#Q1)
numbers = [1, 2, 3, 4, 5]
squares = list(map(lambda x: x * x, numbers))
print("Squares:", squares)
```

```python
#Q2)
names = ["sarika", "rahul", "priya", "amit"]
uppercase_names = list(map(lambda name: name.upper(), names))
print("Uppercase Names:", uppercase_names)
```

```python
#Q3)
prices = [100, 200, 500, 1000]
discounted_prices = list(map(lambda price: price - (price * 0.10), prices))
print("Discounted Prices:", discounted_prices)
```

```python
#Q4)
numbers = [11, 12, 13, 14, 15, 16, 17, 18, 19]
even_numbers = list(filter(lambda x: x % 2 == 0, numbers))
print("Even Numbers:", even_numbers)
```

```python
#Q5)
marks = [25, 40, 32, 55, 70, 30, 45]
passed_students = list(filter(lambda x: x >= 35, marks))
print("Students scoring 35 or more:", passed_students)
```

```python
#Q6)
strings = ["Python", "", "Java", "", "C++", "HTML", ""]
non_empty = list(filter(lambda x: x != "", strings))
print("Non-empty strings:", non_empty)
```

```python
#Q7)
words = ["Python", "Java", "HTML", "Programming"]
lengths = list(map(lambda x: len(x), words))
print("Length of strings:", lengths)
```

```python
#Q8)
from functools import reduce
numbers = [10, 20, 30, 40, 50]
total = reduce(lambda x, y: x + y, numbers)
print("Sum:", total)
```

```python
#Q9)
salaries = [20000, 25000, 30000, 40000]
increased_salaries = list(map(lambda salary: salary + (salary * 0.15), salaries))
print("Increased Salaries:", increased_salaries)
```

```python
#Q10)
numbers = [10, 15, 20, 30, 35, 45, 50, 60, 75]
result = list(filter(lambda x: x % 3 == 0 and x % 5 == 0, numbers))
print("Numbers divisible by both 3 and 5:", result)
```
