```python
# Q1)Write a program to find the largest and smallest element in a tuple without using
numbers = (12, 45, 67, 89, 54)
largest = numbers[0]
smallest = numbers[0]
for num in numbers:
    if num > largest:
        largest = num
    if num < smallest:
        smallest = num
print("Largest element:", largest)
print("Smallest element:", smallest)
```

```python
#Q2)Write a program to count the number of even and odd numbers in a tuple
t = (105, 15, 230, 25, 30, 35)
even = 0
odd = 0
for i in t:
    if i % 2 == 0:
     even += 1
    else:
     odd += 1
print("Even:", even)
print("Odd:", odd)
```

```python
#Q3)Write a program to find all common elements between two tuples
t1 = (170, 260, 530, 340)
t2 = (20, 450, 60)
common = ()
for i in t1:
    if i in t2:
        common += (i,)
print("Common Elements:", common)
```

```python
#Q4)Write a program to create a new tuple containing only positive numbers from agiven tuple..
t = (-10, 20, -5, 40, 0, 15)
positive = ()
for i in t:
    if i > 0:
        positive += (i,)
print("Positive Tuple:", positive)
```

```python
#Q5)Write a program to remove duplicate elements from a tuple while preserving theoriginal order.
t = (10, 237, 30, 20, 44, 140, 540)
result = ()
for i in t:
    if i not in result:
        result += (i,)
print(result)
```

```python
#Q6)Write a program to separate all even and odd elements of a tuple into two differenttuples.
#Input
t = (10, 15, 20, 25, 30)
even = ()
odd = ()
for i in t:
    if i % 2 == 0:
        even += (i,)
    else:
        odd += (i,)
print("Even =", even)
print("Odd =", odd)
```

```python
#Q7)Write a program to find the intersection of two tuples.

t1 = (1, 2, 3, 4)
t2 = (3, 4, 5, 6)
intersection = ()
for i in t1:
    if i in t2:
        intersection += (i,)
print("Intersection:", intersection)
```

```python
#Q8)Write a program to find the union of two tuples (unique elements only)
t1 = (1, 2, 3, 4)
t2 = (3, 4, 5, 6)
union = ()
for i in t1:
    if i not in union:
        union += (i,)
for i in t2:
    if i not in union:
        union += (i,)

print("Union:", union)
```

```python
#Q9)Write a program to convert a nested tuple like:
t = ((1, 2), (3, 4), (5, 6))
result = ()
for i in t:
    for j in i:
        result += (j,)
print(result)
```
